```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
fig, axs = plt.subplots()
x=np.linspace(1, 100)
y=np.sin(x)
axs.plot(x, y)
plt.show()
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Generate x values
x = np.linspace(-10, 10, 100)
y = sigmoid(x)

# Plot
plt.figure(figsize=(8, 5))
plt.plot(x, y, 'b-', linewidth=2)
plt.xlabel('x')
plt.ylabel('σ(x)')
plt.title('Sigmoid Function')
plt.grid(True, alpha=0.3)
plt.axhline(y=0.5, color='r', linestyle='--', alpha=0.5, label='y = 0.5')
plt.axvline(x=0, color='g', linestyle='--', alpha=0.5, label='x = 0')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Simple plot() parameters demonstration
x = np.linspace(0, 10, 50)

# Different colors and styles
plt.plot(x, np.sin(x), 'r-', label='red line')          # red line
plt.plot(x, np.cos(x), 'b--', label='blue dashed')      # blue dashed
plt.plot(x, np.sin(x) * 0.5, 'go', label='green dots')  # green dots

plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('Simple Plot Example')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Plots for multiple functions
x = np.linspace(-2*np.pi, 2*np.pi, 200)

fig, axs = plt.subplots(2, 3, figsize=(15, 10))
fig.suptitle('Common Mathematical Functions', fontsize=16)

# 1. Sine function
axs[0, 0].plot(x, np.sin(x), 'b-', linewidth=2)
axs[0, 0].set_title('Sine Function')
axs[0, 0].grid(True, alpha=0.3)
axs[0, 0].set_xlabel('x')
axs[0, 0].set_ylabel('sin(x)')

# 2. Cosine function
axs[0, 1].plot(x, np.cos(x), 'r-', linewidth=2)
axs[0, 1].set_title('Cosine Function')
axs[0, 1].grid(True, alpha=0.3)
axs[0, 1].set_xlabel('x')
axs[0, 1].set_ylabel('cos(x)')

# 3. Tangent function
axs[0, 2].plot(x, np.tan(x), 'g-', linewidth=2)
axs[0, 2].set_title('Tangent Function')
axs[0, 2].set_ylim(-10, 10)
axs[0, 2].grid(True, alpha=0.3)
axs[0, 2].set_xlabel('x')
axs[0, 2].set_ylabel('tan(x)')

# 4. Exponential function
x_exp = np.linspace(-2, 2, 100)
axs[1, 0].plot(x_exp, np.exp(x_exp), 'm-', linewidth=2)
axs[1, 0].set_title('Exponential Function')
axs[1, 0].grid(True, alpha=0.3)
axs[1, 0].set_xlabel('x')
axs[1, 0].set_ylabel('e^x')

# 5. Logarithm function
x_log = np.linspace(0.1, 5, 100)
axs[1, 1].plot(x_log, np.log(x_log), 'c-', linewidth=2)
axs[1, 1].set_title('Natural Logarithm Function')
axs[1, 1].grid(True, alpha=0.3)
axs[1, 1].set_xlabel('x')
axs[1, 1].set_ylabel('ln(x)')

# 6. Sigmoid function (defined earlier)
x_sig = np.linspace(-10, 10, 100)
axs[1, 2].plot(x_sig, sigmoid(x_sig), 'orange', linewidth=2)
axs[1, 2].set_title('Sigmoid Function')
axs[1, 2].grid(True, alpha=0.3)
axs[1, 2].set_xlabel('x')
axs[1, 2].set_ylabel('σ(x)')

plt.tight_layout()
plt.show()
```

```python
# Create sample 2D feature data
X = np.random.randn(100, 4)  # 100 samples, 4 features
y = np.random.randn(100)  # 100 target values

nrows, ncols = 1, X.shape[1]  # Determine grid size based on number of features
fig, axes = plt.subplots(nrows, ncols, figsize=(15, 4))
axes = axes.flatten()  # Flatten to 1D array
for i in range(X.shape[1]):
    axes[i].scatter(X[:, i], y, alpha=0.5, s=20)
    axes[i].set_title(f'Feature [{i}] vs Target')
    axes[i].set_xlabel(f"Feature [{i}]")
    axes[i].set_ylabel("Target")

plt.tight_layout()
plt.show()
```

```python
X = np.random.randn(100, 10)
y = np.random.randn(100)

fig, axes = plt.subplots(nrows=2, ncols=5, figsize=(15, 4), sharey=True)
axes = axes.flatten()
for i in range(10):
    axes[i].scatter(X[:, i], y, alpha=0.5, s=20)
    axes[i].set_title(f'Feature {i+1} vs Target')
    axes[i].set_xlabel(f"Feature {i+1}")
    if i % 5 == 0:
        axes[i].set_ylabel("Target y")

plt.tight_layout()
plt.show()
```
